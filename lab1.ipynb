```python
def show_image(img, title="", cmap=None):
    plt.figure(figsize=(5,5))
    plt.title(title)
    plt.axis("off")

    if len(img.shape) == 3 and cmap is None:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    else:
        plt.imshow(img, cmap=cmap)

    plt.show()
```

## Task 3 – Smanjenje rezolucije uzorkovanjem svakog drugog piksela

```python
img = cv2.imread("/content/airplane.jpg")

if img is None:
    print("Greška: slika nije učitana.")
else:

    half_vertical = img[::2, :, :]


    half_horizontal = img[:, ::2, :]


    half_both = img[::2, ::2, :]

    plt.figure(figsize=(15,5))

    plt.subplot(1,3,1)
    plt.title(f"Half vertical\n{half_vertical.shape}")
    plt.axis("off")
    plt.imshow(cv2.cvtColor(half_vertical, cv2.COLOR_BGR2RGB))

    plt.subplot(1,3,2)
    plt.title(f"Half horizontal\n{half_horizontal.shape}")
    plt.axis("off")
    plt.imshow(cv2.cvtColor(half_horizontal, cv2.COLOR_BGR2RGB))

    plt.subplot(1,3,3)
    plt.title(f"Half both\n{half_both.shape}")
    plt.axis("off")
    plt.imshow(cv2.cvtColor(half_both, cv2.COLOR_BGR2RGB))

    plt.tight_layout()
    plt.show()
```

## Task 4 – Promjena Hue komponente u HSV prostoru za +30

```python
img = cv2.imread("/content/airplane.jpg")

if img is None:
    print("Greška: slika nije učitana.")
else:
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)


    hsv[:, :, 0] = (hsv[:, :, 0].astype(np.int16) + 30) % 180

    result = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

    plt.figure(figsize=(6,6))
    plt.title("HSV - H + 30")
    plt.axis("off")
    plt.imshow(result)
    plt.show()
```

## Task 5 – Postavljanje Hue komponente na 0

```python
img = cv2.imread("/content/airplane.jpg")

if img is None:
    print("Greška: slika nije učitana.")
else:
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)


    hsv[:, :, 0] = 0

    result = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

    plt.figure(figsize=(6,6))
    plt.title("HSV - H = 0")
    plt.axis("off")
    plt.imshow(result)
    plt.show()
```
